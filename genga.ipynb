# Check the compute capability of the GPU

Before we can compile GENGA, we must know which compute capability the available GPU has. In order to determine this, compile and execute the provided helper program CheckGPU:

```python
!nvcc -o CheckGPU CheckGPU.cu
!./CheckGPU
```

# Compile GENGA

We have determined the compute capability of the available GPU in the last step. For example, if we have a Tesla K80 GPU, then the compute capability would be 37. 

We now use this compute capability to compile GENGA with the following command. Please update the number (37) with your detected compute capability.

```python
!make SM=37
```

# Create a test simulation directory

We have now compiled GENGA and got a code executable named 'genga'.
Before we run the code, we create a new directory, containing the default initial conditions, a user parameter file and the code executable.

Run the following command:

```python
%cd /content
!mkdir GengaTest
%cd /content/GengaTest
%cp ../genga/source/genga .
%cp ../genga/source/param.dat .
%cp ../genga/source/initialplanet_2048.dat .
!ls
```

# Run GENGA

Now, we are ready to run GENGA. The default user parameters in the param.dat file will run the code for 100000 time steps.

If you want to change the number of time steps or other parameters, then
modify the param.dat file in this directory, or use GENGA console arguments.
Please see in the GENGA documentation for detailed information.

Note that in colab, you can stop a running code box by clicking on the stop button.

```python
!./genga
```

When GENGA is stopped manually, then the code automatically saves the latest reached time step. 

GENGA can then be restarted from that latest time step with:

```python
!./genga -R -1
```

If you want, you can delete the generated output files with:

```python
!rm Out*
```

#Run GENGA with timeout

Sometimes, it is useful to set a specific time that GENGA is allowed to run.
When GENGA is running on a computer cluster with a job manager (e.g slurm) then this can be set with the manager options. Sending a SIGINT signal 2 minutes before the end of the run time allows GENGA to stop savely and to store the latest time step, before the code session gets killed. 

If not a jog manager is used, then the same behavior can be set up with the 'timeout' command. With the 'timeout' option, it is possible to send a SIGINT signal to GENGA after a specific time. For example:

!timeout -s SIGINT 20 ./genga : stops GENGA after 20 seconds

!timeout -s SIGINT 5m ./genga : stops GENGA after 5 minutes

!timeout -s SIGINT 2h ./genga : stops GENGA after 2 hours

Try it with

```python
!timeout -s SIGINT 20 ./genga
```
